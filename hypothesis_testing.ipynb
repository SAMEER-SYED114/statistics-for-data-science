z- test using p value


```python
import numpy as np
from scipy.stats import norm
sample=[172,174,168,169,171,173,175,170,169,172]
('using sample only few for practice generally perform this when sample is above 30 ')
population_mean=170
population_std = 3
samp_mean=np.mean(sample)
count=len(sample)
alpha=0.05
```

```python
z_score=(samp_mean-population_mean)/(population_std/np.sqrt(count))
z_score
```

```python
p_value=2 * (1-norm.cdf(abs(z_score)))
p_value
```

```python
if p_value < alpha:
    print('i wil reject null_hypothesis')
else:
    print('fail to reject null hypothesis')
```

t-test

```python
#if we have sample standard deviation and sample <30 we will perfrom t test
#in some cases if we have sample above 30 but we only have sample std we will perfrom t test
```

```python
import numpy as np
from scipy import stats
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
samp_mean=np.mean(sample)
sample_std = np.std(sample,ddof = 1)
n=len(sample)
mean_population=170
alpha=0.05
```

```python
t_test=(samp_mean-mean_population)/(sample_std/np.sqrt(n))
t_test
```

```python

p_value = 2 * (1 - stats.t.cdf(abs(t_test),df = n -1 ))
if p_value<alpha:
    print('reject null hypothesis')
else:
    print('fail to reject null hypothesis')
```

TWO- SAMPLE -TEST

```python
import numpy as np
from scipy import stats

group_A = [85, 88, 90, 92, 87, 85, 89, 91, 86, 88] 
group_B = [82, 84, 80, 83, 81, 79, 78, 85, 84, 83]  
alpha=0.05

T_test,P_value=stats.ttest_ind(group_A,group_B,equal_var=False)

P_value
```

```python
if P_value<alpha:
    print('reject null hypothesis')
else:
    print('fail to reject null hypothesis')
```

CHI- SQUARE TEST

```python
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import chi2_contingency
data_set=sns.load_dataset('titanic')
data_set
```

```python
contingency_table=pd.crosstab(data_set['sex'],data_set['survived'])
contingency_table
```

```python
chi2,p__value,dof,expected=chi2_contingency(contingency_table)

chi2
```

```python
p__value
```

```python
dof
alpha=0.05
```

```python
if p__value < alpha:
    print('reject H0: survival depends on sex')
else:
    print('fail to reject H0: no evidence survival depends on sex')
```

ANNOVA TEST

```python
import pandas as pd
import seaborn as sns
from scipy.stats import f_oneway

df=sns.load_dataset('titanic')
df
```

```python
df=df[['age','pclass']].dropna()
```

```python

df['pclass'].unique()
```

```python



class_1 = df[df['pclass'] == 1]['age']
class_2 = df[df['pclass'] == 2]['age']
class_3 = df[df['pclass'] == 3]['age']
class_1

```

```python
f_stat,p_value=f_oneway(class_1,class_2,class_3)
```

```python
f_stat
```

```python
alpha=0.05
p_value
```

```python
if p_value < alpha:
    print('i will reject null hypothsis ')
else:
    print('i will accept null pypothsis')
```

DOT PRODUCT

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, 1])
b = np.array([1, 3])

plt.quiver(0, 0, *a, angles="xy", scale_units="xy", scale=1, color="blue")
plt.quiver(0, 0, *b, angles="xy", scale_units="xy", scale=1, color="red")
plt.xlim(0, 4)
plt.ylim(0, 4)
plt.title(f"Dot product = {np.dot(a, b)}")
plt.show()
```

CROSS PRODUCT

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([1, 0, 0])
b = np.array([0, 1, 0])
c = np.cross(a, b)

ax = plt.figure().add_subplot(projection="3d")
ax.quiver(0, 0, 0, *a, color="blue")
ax.quiver(0, 0, 0, *b, color="red")
ax.quiver(0, 0, 0, *c, color="green")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_zlim(0, 1)
plt.show()
```

PROJECTION

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, 4])
b = np.array([5, 0])
p = (np.dot(a, b) / np.dot(b, b)) * b

plt.quiver(0, 0, *a, angles="xy", scale_units="xy", scale=1, color="blue")
plt.quiver(0, 0, *b, angles="xy", scale_units="xy", scale=1, color="red")
plt.quiver(0, 0, *p, angles="xy", scale_units="xy", scale=1, color="green")
plt.xlim(0, 6)
plt.ylim(0, 6)
plt.show()
```
